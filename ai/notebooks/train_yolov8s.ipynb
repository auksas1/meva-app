```python
from ultralytics import YOLO

model = YOLO("yolov8s.pt")

results = model.train(
    data="/content/Vehicle-Damage-Detection-1/data.yaml",
    epochs=30,
    imgsz=640,
    batch=16,
    project="/content/runs",
    name="vehicle_damage_yolov8s_30e"
)
```

```python
from ultralytics import YOLO

best_model = YOLO("/content/runs/vehicle_damage_yolov8s_30e/weights/best.pt")

best_model.predict(
    source="/content/Vehicle-Damage-Detection-1/valid/images",
    save=True,
    conf=0.25,
    project="/content/predict_runs",
    name="vehicle_damage_preds_yolov8s_30e"
)
```
